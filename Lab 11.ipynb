## **Part 1: What is Streamlit?**

Streamlit is an open-source Python library that turns **plain Python scripts** into **interactive web apps** — with no HTML, CSS, or JavaScript required.

### Why Streamlit?
| Feature | Benefit |
|---------|---------|
| Pure Python | You already know the language |
| Automatic UI | Widgets, layout, and styling built in |
| Live reload | Edit code → save → app updates instantly |
| Easy deployment | One-click deploy via Streamlit Community Cloud |
| Data-friendly | First-class support for pandas, matplotlib, Plotly, etc. |

### The Streamlit Mental Model
```
streamlit_app.py  →  streamlit run app.py  →  browser at localhost:8501
```

Every time a user interacts with a widget, **the entire script re-runs from top to bottom**. This is the single most important concept to understand.

> **Key idea:** Streamlit is not a traditional web framework with routes and callbacks. It is a **script that redraws itself** on every interaction.

## **Part 2: Why You Must Read the Docs**

Streamlit evolves quickly. New widgets and parameters are added regularly, and function signatures change.

### How to Read a Streamlit Doc Page
Every API page shows:
1. **Function signature** — the parameters and their defaults.
2. **Parameters table** — what each argument does.
3. **Example code** — copy-paste ready.
4. **Version notes** — when a feature was added.

> **Best practice:** Before asking "how do I do X?", search the docs. The answer is almost always there with a runnable example.

### Common Doc-Search Habits
- Search `site:docs.streamlit.io <keyword>` in Google.
- Use the docs' built-in search bar.
- Check the **Cheat Sheet** for a quick list of every widget.
- Read the **version-added** note if a tutorial looks out of date.

## **Part 4: Core Streamlit Methods**

Below are the functions you will use most often. **Always confirm the signature in the docs** — some parameters are new.

### Text & Display
```python
st.title("My App")
st.header("Section")
st.subheader("Subsection")
st.write("Anything: text, numbers, dataframes, plots")
st.markdown("**bold** and *italic*")
st.code("print('hello')", language="python")
```

### Widgets (Inputs)
```python
st.text_input("Your name")
st.number_input("Age", min_value=0, max_value=120)
st.slider("Pick a number", 0, 100, 50)
st.selectbox("Choose", ["A", "B", "C"])
st.checkbox("I agree")
st.button("Submit")
```

### Layout
```python
col1, col2 = st.columns(2)
with st.sidebar:
    st.write("Sidebar content")
st.tabs(["Tab 1", "Tab 2"])
```

### Data Display
```python
st.dataframe(df)
st.table(df)
st.metric("Revenue", "$1,200", "+5%")
```

### Feedback
```python
st.success("Done!")
st.error("Something went wrong")
st.warning("Careful")
st.info("For your information")
```

## **Part 5: Build a Basic App from Scratch**

We will build a simple **Greeting + Calculator** app that demonstrates:
- A sidebar with an input.
- A main area with two tabs: **Greeting** and **Calculator**.
- Basic widgets and layout.

### The Full Script (`app.py`)
```python
import streamlit as st

st.set_page_config(page_title="My First App", page_icon="✨")

# Sidebar
with st.sidebar:
    st.header("Settings")
    name = st.text_input("Your name", "Student")

# Main title
st.title("Welcome to My First Streamlit App")
st.write(f"Hello, **{name}**! This app is built in pure Python.")

# Tabs
tab1, tab2 = st.tabs(["Greeting", "Calculator"])

with tab1:
    st.subheader("A Friendly Greeting")
    mood = st.selectbox("How are you feeling?", ["Happy", "Neutral", "Tired"])
    if st.button("Say hello"):
        st.success(f"Great to hear you're {mood.lower()}, {name}!")

with tab2:
    st.subheader("Simple Calculator")
    col1, col2 = st.columns(2)
    with col1:
        a = st.number_input("First number", value=0.0)
    with col2:
        b = st.number_input("Second number", value=0.0)
    op = st.radio("Operation", ["+", "-", "×", "÷"], horizontal=True)
    if st.button("Calculate"):
        if op == "+":
            result = a + b
        elif op == "-":
            result = a - b
        elif op == "×":
            result = a * b
        else:
            result = a / b if b != 0 else "undefined (division by zero)"
        st.metric("Result", result)
```

### What to Notice
- `st.set_page_config` must be the **first** Streamlit call.
- `with st.sidebar:` puts widgets on the left panel.
- `st.tabs` organizes the main area into clickable tabs.
- `st.columns` lays widgets side by side.
- Every widget returns a value you can use in `if` statements.
- The whole script re-runs on every button click, so state resets unless you use `st.session_state` (more on that later).

````python
# Example 1: Write the full app to app.py so you can run it in the terminal
app_code = '''import streamlit as st

st.set_page_config(page_title="My First App", page_icon="✨")

with st.sidebar:
    st.header("Settings")
    name = st.text_input("Your name", "Student")

st.title("Welcome to My First Streamlit App")
st.write(f"Hello, **{name}**! This app is built in pure Python.")

tab1, tab2 = st.tabs(["Greeting", "Calculator"])

with tab1:
    st.subheader("A Friendly Greeting")
    mood = st.selectbox("How are you feeling?", ["Happy", "Neutral", "Tired"])
    if st.button("Say hello"):
        st.success(f"Great to hear you're {mood.lower()}, {name}!")

with tab2:
    st.subheader("Simple Calculator")
    col1, col2 = st.columns(2)
    with col1:
        a = st.number_input("First number", value=0.0)
    with col2:
        b = st.number_input("Second number", value=0.0)
    op = st.radio("Operation", ["+", "-", "×", "÷"], horizontal=True)
    if st.button("Calculate"):
        if op == "+":
            result = a + b
        elif op == "-":
            result = a - b
        elif op == "×":
            result = a * b
        else:
            result = a / b if b != 0 else "undefined (division by zero)"
        st.metric("Result", result)
'''

with open("app.py", "w", encoding="utf-8") as f:
    f.write(app_code)

print("Written app.py — now run: streamlit run app.py")
````

## **Part 6: Deploying to GitHub + Streamlit Community Cloud**

You already learned how to connect GitHub to Streamlit Community Cloud. Here is the **minimum project structure** you need.

### Project Structure
```
streamlit-demo/
├── app.py
├── requirements.txt
└── .gitignore
```

### `requirements.txt`
```
streamlit
```
Add any other libraries your app imports (e.g., `pandas`, `matplotlib`).

### `.gitignore`
```
__pycache__/
*.pyc
.venv/
.DS_Store
```

### Deploy Steps (recap)
1. Create a **public GitHub repo** and push `app.py`, `requirements.txt`, `.gitignore`.
2. Go to https://share.streamlit.io and sign in with GitHub.
3. Click **New app**, pick your repo, branch, and main file (`app.py`).
4. Click **Deploy**. Wait ~1–2 minutes.
5. Share the generated URL (e.g., `https://your-app.streamlit.app`).

> **Tip:** If deployment fails, read the **build log** — it almost always names the missing package or the line of code that crashed.

````python
# Example 2: Create the supporting files for deployment
with open("requirements.txt", "w", encoding="utf-8") as f:
    f.write("streamlit\n")

with open(".gitignore", "w", encoding="utf-8") as f:
    f.write("__pycache__/\n*.pyc\n.venv/\n.DS_Store\n")

print("Created requirements.txt and .gitignore")
print("Next: git init && git add . && git commit -m 'first app' && git push")
````
